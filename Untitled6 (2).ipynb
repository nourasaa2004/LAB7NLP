## Task 1 — Model Families

1. Sentiment classification: Encoder-only
2. Machine translation: Encoder–decoder
3. Open-ended text generation: Decoder-only
4. Named Entity Recognition: Encoder-only
5. Text summarization: Encoder–decoder

```python

from transformers import pipeline

# Load sentiment analysis model
classifier = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

# Sentences
sentences = [
    "The movie was excellent.",
    "The lecture was difficult to understand.",
    "I enjoyed learning about transformers."
]

# Predict sentiment
for sentence in sentences:
    result = classifier(sentence)
    print("Sentence:", sentence)
    print("Label:", result[0]["label"])
    print("Score:", result[0]["score"])
    print()
```

```python

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Load FLAN-T5 model
tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")

# Short paragraph
paragraph = """
Artificial intelligence is changing the way people live and work.
It helps computers understand language, recognize images, and
solve problems. AI is widely used in healthcare, education,
and transportation to improve efficiency and accuracy.
"""

# Summarize the paragraph
prompt = "Summarize the following paragraph: " + paragraph

inputs = tokenizer(prompt, return_tensors="pt")

output = model.generate(**inputs, max_new_tokens=50)

summary = tokenizer.decode(output[0], skip_special_tokens=True)

print("Original Paragraph:", paragraph)
print("Summary:", summary)
```

### Task 5 — Short Reflection

A decoder-only model is suitable for a chatbot like ChatGPT because it generates text by predicting the next token. It uses the previous conversation to understand the context and produce relevant responses. It can answer questions, generate text, and communicate naturally with users. This makes it effective for interactive conversations.

```python

from transformers import pipeline

# Load decoder-only model
generator = pipeline("text-generation", model="distilgpt2")

# Same prompt
prompt = "Artificial intelligence will"

# Generate text multiple times
for i in range(3):
    output = generator(
        prompt,
        max_new_tokens=30,
        do_sample=True,
        pad_token_id=generator.tokenizer.eos_token_id
    )
    print("Output", i + 1, ":", output[0]["generated_text"])
    print()
```

### Optional Challenge — Answer

The outputs can be different because sampling introduces randomness when selecting the next token. Even with the same prompt, the model may choose different words based on their probabilities. This makes the generated text more diverse and creative.
